```python
class MyNumpyEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, np.integer):
            return int(obj)
        elif isinstance(obj, np.floating):
            return float(obj)
        elif isinstance(obj, np.ndarray):
            return obj.tolist()
        else:
            return super(MyEncoder, self).default(obj)


def ensure_dir(file_name):
    os.makedirs(os.path.dirname(file_name), exist_ok=True)
    return file_name


def list_dirs(path=os.getcwd()):
    files = glob.glob(os.path.join(path, '*'))
    files = [x for x in files if os.path.isdir(x)]
    return files


def file_exists(file_name, path_name=os.getcwd()):
    return os.path.isfile(os.path.join(path_name, file_name))


def write_json(json_obj, file_path):
    with open(file_path, 'w') as outfile:
        json.dump(json_obj, outfile, sort_keys=True, indent=4,
                  cls=MyNumpyEncoder)
    return json_obj


def write_np(np_obj, file_path):
    with open(file_path, 'wb') as outfile:
        np.save(outfile, np_obj)
```

## Extracting data

---

```python
hcpya_dir = '/mountpoint/data/HCP_1200/'
hcpya_subjects = [x.split('/')[-1] for x in list_dirs(hcpya_dir)]
len(hcpya_subjects)
```

```python
items = [
    "lh.white", "rh.white",
    "lh.pial", "rh.pial",
    "lh.thickness", "rh.thickness",
    "lh.orig.nofix", "rh.orig.nofix",
    "lh.sphere.reg", "rh.sphere.reg",
]
hcpya_valid_subjects = [
    subject for subject in hcpya_subjects
    if all([
        file_exists(f'/mountpoint/data/HCP_1200/{subject}/T1w/{subject}/surf/{item}','')
        for item in items
    ])
]
len(hcpya_valid_subjects)
```

```python
# ignore warning
nib.imageglobals.logger.setLevel(40)
```

```python
# Store high-resolution thickness for each individual in a separate file
for idx, subject in enumerate(tqdm(hcpya_valid_subjects)):
    sub_dir = f"{idx:02d}"[-2:]
    freesurfer_directory = f"/mountpoint/data/HCP_1200/{subject}/T1w/{subject}/"
    transformed_fslr_thickness = snm.utils.nitools.compute_fslr_thickness(freesurfer_directory)
    np.save(
        ensure_dir(f"/mountpoint/data/normative/fs_LR_32k/HCP-YA/{sub_dir}/{subject}.thickness.fslr.npy"),
        transformed_fslr_thickness.astype(np.float32)
    )
```

```python
unrestricted = pd.read_csv(f'{hcpya_dir}/unrestricted_sination_5_23_2019_21_35_35.csv')
restricted = pd.read_csv(f'{hcpya_dir}/0.RESTRICTED_sination_5_23_2019_21_32_45.csv')
```

```python
# unrestricted[["MRsession_Scanner_3T", 'MRsession_Scans_3T', 'MRsession_Label_3T',]]
unrestricted["MRsession_Scanner_3T"].unique()
```

```python
hcpya_ages = np.array(
    [
        float(restricted[restricted['Subject'] == int(subject)]['Age_in_Yrs'])
        for subject in hcpya_valid_subjects
    ]
)
```

```python
hcpya_genders = np.array(
    [
        unrestricted[unrestricted['Subject'] == int(subject)]['Gender'].copy()
        for subject in hcpya_valid_subjects
    ]
)[:, 0]
hcpya_genders
```

```python
hcpya_ids = np.array(
    [
        str(unrestricted[unrestricted['Subject'] == int(subject)]['Subject'].values[0])
        for subject in hcpya_valid_subjects
    ]
)
hcpya_ids
```

```python
hcpya_enos = np.array(
    [
        float(snm.utils.nitools.compute_total_euler_number(
            f"/mountpoint/data/HCP_1200/{subject}/T1w/{subject}/"
        )) for subject in tqdm(hcpya_valid_subjects)
    ]
)
hcpya_enos
```

```python
hcpya_mean_thickness = np.array(
    [
        np.load(f"/mountpoint/data/normative/fs_LR_32k/HCP-YA/{(idx%100):02d}/{subject}.thickness.fslr.npy").mean()
        for idx, subject in enumerate(tqdm(hcpya_valid_subjects))
    ]
)
hcpya_mean_thickness
```
