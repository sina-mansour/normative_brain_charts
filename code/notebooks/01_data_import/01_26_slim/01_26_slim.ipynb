```python
class MyNumpyEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, np.integer):
            return int(obj)
        elif isinstance(obj, np.floating):
            return float(obj)
        elif isinstance(obj, np.ndarray):
            return obj.tolist()
        else:
            return super(MyEncoder, self).default(obj)


def ensure_dir(file_name):
    os.makedirs(os.path.dirname(file_name), exist_ok=True)
    return file_name


def list_dirs(path=os.getcwd()):
    files = glob.glob(os.path.join(path, '*'))
    files = [x for x in files if os.path.isdir(x)]
    return files


def file_exists(file_name, path_name=os.getcwd()):
    return os.path.isfile(os.path.join(path_name, file_name))


def write_json(json_obj, file_path):
    with open(file_path, 'w') as outfile:
        json.dump(json_obj, outfile, sort_keys=True, indent=4,
                  cls=MyNumpyEncoder)
    return json_obj


def load_json(file_path):
    with open(file_path, 'r') as infile:
        return json.load(infile)


def write_np(np_obj, file_path):
    with open(file_path, 'wb') as outfile:
        np.save(outfile, np_obj)
```

```python
data_info_df_1 = pd.read_csv("/home/sina/storage/Normative_Modeling/data/csv/slim/swu_slim_phenodata_time1.tsv", sep="\t", encoding="latin1")
data_info_df_1["Scan_Time"] = data_info_df_1["Scan_Time"].str.strip("'")

data_info_df_2 = pd.read_csv("/home/sina/storage/Normative_Modeling/data/csv/slim/swu_slim_phenodata_time2.tsv", sep="\t", encoding="latin1")
data_info_df_2["Scan_Time"] = data_info_df_2["Scan_Time"].str.strip("'")

data_info_df_3 = pd.read_csv("/home/sina/storage/Normative_Modeling/data/csv/slim/swu_slim_phenodata_time3.tsv", sep="\t", encoding="latin1")
data_info_df_3["Scan_Time"] = data_info_df_1["Scan_Time"].str.strip("'")

for df in [data_info_df_1, data_info_df_2, data_info_df_3]:
    df["Scan_Time"] = pd.to_datetime(df["Scan_Time"].str.strip("'"), format="%d-%b-%Y")

baseline = data_info_df_1[["Match_ID", "Scan_Time", "age_at_Scan1"]].rename(
    columns={"Scan_Time": "Scan_Time1"}
)

# Merge time2 with baseline
data_info_df_2 = data_info_df_2.merge(
    baseline, on="Match_ID", how="left"
)

# Merge time3 with baseline
data_info_df_3 = data_info_df_3.merge(
    baseline, on="Match_ID", how="left"
)

def compute_age_at_followup(row):
    delta_years = (row["Scan_Time"] - row["Scan_Time1"]).days / 365.25
    return row["age_at_Scan1"] + delta_years

data_info_df_1["age_at_Scan"] = data_info_df_1["age_at_Scan1"]
data_info_df_2["age_at_Scan"] = data_info_df_2.apply(compute_age_at_followup, axis=1)
data_info_df_3["age_at_Scan"] = data_info_df_3.apply(compute_age_at_followup, axis=1)


data_info_df_1.shape, data_info_df_2.shape, data_info_df_3.shape
```

```python
# keep only rows where the path is safe
data_info_df = data_info_df[data_info_df['path'].apply(safe_exists)].copy()
data_info_df.shape
```

```python
from contextlib import suppress

sex_encoder = {
    'female': 'F',
    'male': 'M'
}

valid_subjects_dict = {}

for _, row in tqdm(data_info_df.iterrows()):
    key = f"sub-{row['Match_ID']}_ses-{row['Session']}"
    recon_all_path = f'{row["path"]}'
    if Path(recon_all_path).exists():
        # with suppress(KeyError): # ignore dictionary misses
        valid_subjects_dict[key] = {
            "unique_id": key,
            "participant_id": row["Match_ID"],
            "session_id": row['Session'],
            "sex": sex_encoder[row["Sex"]],
            "age": row["age_at_Scan"],
            # "site": row["site"],
            "scan_path": recon_all_path,
            # "diagnosis": (row["DX_GROUP"] == 1),
        }

len(valid_subjects_dict), list(valid_subjects_dict.items())[:1]
```

```python
items = [
    "lh.white", "rh.white",
    "lh.pial", "rh.pial",
    "lh.thickness", "rh.thickness",
    "lh.orig.nofix", "rh.orig.nofix",
    "lh.sphere.reg", "rh.sphere.reg",
]

def directory_is_valid(path, items = items):
    return len([f for f in items if (path / f).exists()]) == len(items)
```

```python
# Store high-resolution thickness for each individual in a separate file
for idx, subject in enumerate(tqdm(valid_subjects_dict)):
    sub_dir = f"{idx:02d}"[-2:]

    freesurfer_directory = valid_subjects_dict[subject]["scan_path"]
    
    thickness_fslr_output = f"/home/sina/storage/Normative_Modeling/data/fs_LR_32k/{dataset}/{sub_dir}/{subject}.thickness.fslr.npy"

    if (directory_is_valid(Path(freesurfer_directory) / "surf")) and (not Path(thickness_fslr_output).exists()):
        # Compute fslr thickness
        transformed_fslr_thickness = snm.utils.nitools.compute_fslr_thickness(freesurfer_directory)
        np.save(
            ensure_dir(thickness_fslr_output),
            transformed_fslr_thickness.astype(np.float32)
        )
```

```python
%%time
for idx, subject in enumerate(tqdm(valid_subjects_dict)):
    sub_dir = f"{idx:02d}"[-2:]
    valid_subjects_dict[subject]["subject_index"] = idx
    thickness_fslr_output = f"/home/sina/storage/Normative_Modeling/data/fs_LR_32k/{dataset}/{sub_dir}/{subject}.thickness.fslr.npy"
    if Path(thickness_fslr_output).exists():
        valid_subjects_dict[subject]["thickness"] = np.load(
            thickness_fslr_output,
        ).mean()
    else:
        valid_subjects_dict[subject]["thickness"] = np.nan

len(valid_subjects_dict), list(valid_subjects_dict.items())[:1]
```

```python
eno_items = [
    "lh.orig.nofix", "rh.orig.nofix",
]

# Compute Euler Number
for idx, subject in enumerate(tqdm(valid_subjects_dict)):
    if ("euler_no" not in valid_subjects_dict[subject]) or np.isnan(valid_subjects_dict[subject]["euler_no"]):
        sub_dir = f"{idx:02d}"[-2:]
        freesurfer_directory = valid_subjects_dict[subject]["scan_path"]
        if (directory_is_valid(Path(freesurfer_directory) / "surf", items=eno_items)):
            # Compute euler number
            valid_subjects_dict[subject]["euler_no"] = snm.utils.nitools.compute_total_euler_number(
                Path(freesurfer_directory)
            )
        else:
            valid_subjects_dict[subject]["euler_no"] = np.nan
```

```python
# Validity checks
for idx, subject in enumerate(tqdm(valid_subjects_dict)):
    # if "validity_check" not in valid_subjects_dict[subject]:
    valid_subjects_dict[subject]["validity_check"] = (
        # (valid_subjects_dict[subject]["diagnosis"] == False)  # Exclude those with a diagnosis
        # and
        not np.isnan(valid_subjects_dict[subject]["age"])  # Exclude those missing age data
        and
        not np.isnan(valid_subjects_dict[subject]["thickness"])  # Exclude those missing thickness data
        and
        not np.isnan(valid_subjects_dict[subject]["euler_no"])  # Exclude those missing euler number
    )
```
